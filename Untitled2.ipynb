```python

# ==================================================
# AI PDF CHATBOT USING RAG + GEMINI
# Google Colab Project
# ==================================================

# 1. INSTALL LIBRARIES

!pip install -q -U google-genai pypdf gradio numpy

# 2. IMPORT LIBRARIES

import numpy as np
import gradio as gr

from pypdf import PdfReader
from google import genai
from getpass import getpass

print("Libraries installed successfully!")

# 3. CONNECT GEMINI API

API_KEY = getpass("Enter your Gemini API key: ").strip()

if not API_KEY:
    raise ValueError("Please enter a valid API key.")

client = genai.Client(api_key=API_KEY)

CHAT_MODEL = "gemini-2.5-flash"
EMBED_MODEL = "gemini-embedding-001"

print("Gemini connected successfully!")


# 4. SPLIT TEXT INTO CHUNKS

def split_text(text, chunk_size=800, overlap=150):

    chunks = []

    start = 0

    while start < len(text):

        end = start + chunk_size

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        start += chunk_size - overlap

    return chunks


# 5. CREATE EMBEDDINGS

def create_embeddings(text_list):

    result = client.models.embed_content(
        model=EMBED_MODEL,
        contents=text_list
    )

    return np.array(
        [item.values for item in result.embeddings],
        dtype=np.float32
    )


# 6. PROCESS UPLOADED PDF

def process_pdf(pdf_file):

    if pdf_file is None:
        return "Please upload a PDF.", [], None

    try:

        reader = PdfReader(pdf_file)

        text = ""

        for page in reader.pages:

            page_text = page.extract_text()

            if page_text:
                text += page_text + "\n"

        if not text.strip():
            return "No readable PDF text found.", [], None

        chunks = split_text(text)

        if not chunks:
            return "No text chunks found.", [], None

        all_embeddings = []

        for i in range(0, len(chunks), 32):

            batch = chunks[i:i+32]

            vectors = create_embeddings(batch)

            all_embeddings.extend(vectors)

        embeddings = np.array(
            all_embeddings,
            dtype=np.float32
        )

        norms = np.linalg.norm(
            embeddings,
            axis=1,
            keepdims=True
        )

        embeddings = embeddings / np.maximum(norms, 1e-10)

        status = (
            "PDF processed successfully!\n"
            f"Total pages: {len(reader.pages)}\n"
            f"Total chunks: {len(chunks)}\n"
            "Embeddings created successfully!\n"
            "Your chatbot is ready!"
        )

        return status, chunks, embeddings

    except Exception as e:

        return "PDF processing error: " + str(e), [], None


# 7. RETRIEVE RELEVANT INFORMATION

def retrieve_information(question, chunks, embeddings):

    question_vector = create_embeddings([question])[0]

    question_vector = question_vector / max(
        np.linalg.norm(question_vector),
        1e-10
    )

    scores = np.dot(embeddings, question_vector)

    best_indices = np.argsort(scores)[-4:][::-1]

    return [chunks[i] for i in best_indices]


# 8. GENERATE ANSWERS

def ask_question(question, chunks, embeddings):

    if not question.strip():
        return "Please enter your question."

    if not chunks or embeddings is None:
        return "Please upload and process your PDF first."

    try:

        relevant_chunks = retrieve_information(
            question,
            chunks,
            embeddings
        )

        context = "\n\n".join(relevant_chunks)

        prompt = f"""
        You are an intelligent PDF chatbot.

        Answer the question using the PDF context.

        Give simple, clear and accurate answers.

        If the answer is not available in the PDF,
        say that it is not available in the document.

        PDF CONTEXT:
        {context}

        QUESTION:
        {question}

        ANSWER:
        """

        response = client.models.generate_content(
            model=CHAT_MODEL,
            contents=prompt
        )

        return response.text or "No answer generated."

    except Exception as e:

        return "AI Error: " + str(e)


# 9. CREATE CHATBOT WEBSITE

with gr.Blocks(
    title="AI PDF Chatbot",
    theme=gr.themes.Soft()
) as demo:

    gr.Markdown(
        """
        # AI PDF CHATBOT

        ### Ask questions about your PDF using RAG and Gemini AI.

        Upload a PDF and start chatting!
        """
    )

    chunks_state = gr.State([])
    embeddings_state = gr.State(None)

    pdf_input = gr.File(
        label="Upload your PDF",
        file_types=[".pdf"],
        type="filepath"
    )

    process_button = gr.Button(
        "Process PDF",
        variant="primary"
    )

    status = gr.Textbox(
        label="Processing Status",
        lines=4,
        interactive=False
    )

    process_button.click(
        fn=process_pdf,
        inputs=pdf_input,
        outputs=[
            status,
            chunks_state,
            embeddings_state
        ]
    )

    gr.Markdown("## Chat with your PDF")

    question = gr.Textbox(
        label="Ask your question",
        placeholder="What is this PDF about?"
    )

    ask_button = gr.Button(
        "Ask AI",
        variant="primary"
    )

    answer = gr.Markdown()

    ask_button.click(
        fn=ask_question,
        inputs=[
            question,
            chunks_state,
            embeddings_state
        ],
        outputs=answer
    )

    question.submit(
        fn=ask_question,
        inputs=[
            question,
            chunks_state,
            embeddings_state
        ],
        outputs=answer
    )

    gr.Markdown(
        """
        ---
        Built with Python, RAG, Gemini AI and Gradio.
        """
    )


# 10. LAUNCH CHATBOT

print("Launching your chatbot...")

demo.launch(share=True)
```
